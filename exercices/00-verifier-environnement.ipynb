## Un premier aperçu du module

On simule 2 000 réservations, chacune annulée avec une probabilité de 0,37,
et on suit la part des réservations annulées au fur et à mesure. Au début,
elle varie beaucoup ; ensuite, elle se stabilise. C'est l'idée de la
probabilité comme fréquence, vue en séance 1.

```python
rng = np.random.default_rng(7)
annulee = rng.random(2000) < 0.37
frequence = annulee.cumsum() / np.arange(1, annulee.size + 1)

fig, ax = plt.subplots(figsize=(8, 3.5))
ax.plot(np.arange(1, annulee.size + 1), frequence, label="part des réservations annulées")
ax.axhline(0.37, color="grey", linestyle="--", label="probabilité 0,37")
ax.set_xlabel("nombre de réservations observées")
ax.set_ylabel("fréquence")
ax.set_ylim(0, 1)
ax.legend()
plt.show()
```
